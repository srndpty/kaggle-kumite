```python
from pathlib import Path

import pandas as pd

root = Path("/kaggle/input")
paths = [
    path / "isic-2024-challenge" / "sample_submission.csv"
    for path in (root, root / "competitions")
    if (path / "isic-2024-challenge" / "sample_submission.csv").is_file()
]
assert len(paths) == 1
sample = pd.read_csv(paths[0])
assert list(sample.columns) == ["isic_id", "target"]
assert len(sample) and sample.isic_id.notna().all() and sample.isic_id.is_unique
sample.assign(target=0.5).to_csv("submission.csv", index=False)
```
